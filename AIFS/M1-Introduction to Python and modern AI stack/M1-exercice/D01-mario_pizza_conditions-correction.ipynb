## Tâche 2 : Choisir la bonne taille de pizza

La boutique de Mario vend trois tailles basées sur le diamètre en centimètres :
- moins de 25 cm : **Small** (Petite)
- entre 25 et 32 cm : **Medium** (Moyenne)
- 33 cm ou plus : **Large** (Grande)

Écrivez un code qui lit la variable `diameter` et affiche le nom de la taille. Utilisez `if`, `elif` et `else`. Testez avec : **22**, **28** et **35**.

> **Correction :** utilisez `if`, `elif`, `else` dans l'ordre. Python s'arrête à la première condition vraie.

**Sortie attendue :**

```
Pizza size: Medium
```

````python
diameter = 45

if diameter < 25:
    size = "Small"
elif diameter < 33:
    size = "Medium"
else:
    size = "Large"

print("Pizza size:", size)
````

````python
# Test all three values
test_diameters = [22, 28, 35,55,10,26,33]

for diameter in test_diameters:
    if diameter < 25:
        size = "Small"
    elif diameter < 33:
        size = "Medium"
    else:
        size = "Large"
    print("Diameter", diameter, "cm ->", size)
````

````python
# Test the three values
menu_toppings = ["mushroom", "pepperoni", "olive", "onion", "pepper", "ham"]
test_toppings = ["mushroom", "pineapple", "ham"]

for topping in test_toppings:
    if topping in menu_toppings:
        print("Adding", requested_topping, "to the pizza.")
    else:
        print("Sorry, we don't have", requested_topping ,".")
````

## Tâche 4 : Pouvons-nous livrer cette pizza ?

Mario livre uniquement lorsque deux choses sont vraies en même temps : la distance doit être de **5 km ou moins**, et le livreur doit être **disponible**.

```python
distance_km = 3
driver_free = True
```

Affichez `"Delivery confirmed."` uniquement si les deux conditions sont vraies. Sinon, affichez `"Pickup only."`.

Utilisez l'opérateur `and`. Testez ces trois cas : distance 3 avec livreur libre, distance 7 avec livreur libre, distance 3 avec livreur occupé (`False`).

>  **Correction :** `and` combine deux conditions. Les deux doivent être vraies.

````python
distance_km = 3
driver_free = True

if distance_km <= 5 and driver_free:
    print("Delivery confirmed.")
else:
    print("Pickup only.")
````

````python
# Test the three cases
test_distances = [3, 7, 3]
test_drivers = [True, True, False]

for i in range(3): # 0=> 3, 1=> 7 ,2=> 3 
    distance_km = test_distances[i]
    driver_free = test_drivers[i]
    print("Distance:", distance_km, "km, Driver free:", driver_free)
    if distance_km <= 5 and driver_free:
        print("  -> Delivery confirmed.")
    else:
        print("  -> Pickup only.")
````

````python
objects = [
{"distance_km" : 3, "driver_free": True},
{"distance_km" : 7, "driver_free": True}, 
{"distance_km" : 3, "driver_free": False}
]
for object in objects:

    print("Distance:" ,{object["distance_km"]}, "km, Driver free: ",{object['driver_free']})

    if object["distance_km"] <= 5 and object["driver_free"] == True:
        print(" -> Delivery confirmed.") 
    else:
        print(" -> Pickup only.")
````

## Tâche 8 : La vérification finale de la commande

Cette tâche rassemble tout. Mario vous donne la commande complète :

```python
pizza_size = "Medium"
topping = "pepperoni"
distance_km = 4
driver_free = True
menu_toppings = ["mushroom", "pepperoni", "olive", "onion", "pepper", "ham"]
```

Votre code doit vérifier trois choses **dans l'ordre** :
1. La garniture est-elle au menu ? Si non, affichez `"Order refused: topping not available."` et arrêtez-vous.
2. La livraison est-elle possible (distance ≤ 5 km **et** chauffeur disponible) ? Affichez `"Delivery OK"` ou `"Pickup only"`.
3. Affichez une confirmation finale telle que `"Medium pepperoni pizza confirmed. Delivery OK."`.

> **Correction :** la solution enchaîne les trois vérifications. La vérification du menu s'exécute en premier. Si la garniture est invalide, le code s'arrête grâce à des `if`/`else` imbriqués (équivalent d'une protection de type `return`).

````python
pizza_size = "Medium"
topping = "pepperoni"
distance_km = 8
driver_free = True
menu_toppings = ["mushroom", "pepperoni", "olive", "onion", "pepper", "ham"]

# Step 1: check if topping is on the menu
if topping not in menu_toppings:
    print("Order refused: topping not available.")
else:
    # Step 2: check delivery
    if distance_km <= 5 and driver_free:
        delivery_status = "Delivery OK"
    else:
        delivery_status = "Pickup only"
    print(delivery_status)

    # Step 3: final confirmation
    print(pizza_size, topping, "pizza confirmed.", delivery_status , ".")
````
